```python
# ==========================================
# EXERCISE 1 — DESCRIPTIVE ANALYSIS
# Datasets:
# 1. Titanic
# 7. Spotify Songs
# 8. World Happiness Report
# ==========================================

# ------------------------------------------
# STEP 1: IMPORT LIBRARIES
# ------------------------------------------

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: make plots look nicer
sns.set_style("whitegrid")

# ==========================================
# PART A — TITANIC DATASET
# ==========================================

print("=" * 50)
print("TITANIC DATASET")
print("=" * 50)

# Load Titanic dataset
titanic = pd.read_csv("/content/Titanic-Dataset.csv")

# ------------------------------------------
# TASK 1
# Display first 10 rows
# ------------------------------------------

print("\nFIRST 10 ROWS:")
print(titanic.head(10))

# ------------------------------------------
# Display Data Types
# ------------------------------------------

print("\nDATA TYPES:")
print(titanic.dtypes)

# ------------------------------------------
# Display Shape
# ------------------------------------------

print("\nDATASET SHAPE:")
print(titanic.shape)

# ==========================================
# TASK 2
# Descriptive Statistics
# ==========================================

print("\nDESCRIPTIVE STATISTICS")

numeric_cols = titanic.select_dtypes(include=['number'])

print("\nMEAN")
print(numeric_cols.mean())

print("\nMEDIAN")
print(numeric_cols.median())

print("\nSTANDARD DEVIATION")
print(numeric_cols.std())

print("\nMINIMUM")
print(numeric_cols.min())

print("\nMAXIMUM")
print(numeric_cols.max())

print("\nSKEWNESS")
print(numeric_cols.skew())

print("\nKURTOSIS")
print(numeric_cols.kurt())

# ==========================================
# TASK 3
# Histogram of Passenger Ages
# ==========================================

plt.figure(figsize=(8,5))

titanic['Age'].hist(
    bins=20,
    edgecolor='black'
)

plt.title("Titanic Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()

# ==========================================
# TASK 4
# Survival Count by Passenger Class
# ==========================================

plt.figure(figsize=(8,5))

survival_by_class = titanic.groupby('Pclass')['Survived'].sum()

survival_by_class.plot(
    kind='bar'
)

plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Survivors")

plt.show()

# ==========================================
# PART B — SPOTIFY DATASET
# =========================================
# ==========================================
# END OF EXERCISE 1
# ==========================================
```
