```python
# ==========================================
# EXERCISE 1 — DESCRIPTIVE ANALYSIS
# Datasets:
# 1. Titanic
# 7. Spotify Songs
# 8. World Happiness Report
# ==========================================

# ------------------------------------------
# STEP 1: IMPORT LIBRARIES
# ------------------------------------------

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: make plots look nicer
sns.set_style("whitegrid")

# ==========================================
# PART A — TITANIC DATASET
# ==========================================

print("=" * 50)
print("TITANIC DATASET")
print("=" * 50)

# Load Titanic dataset
titanic = pd.read_csv("/content/Titanic-Dataset.csv")

# ------------------------------------------
# TASK 1
# Display first 10 rows
# ------------------------------------------

print("\nFIRST 10 ROWS:")
print(titanic.head(10))

# ------------------------------------------
# Display Data Types
# ------------------------------------------

print("\nDATA TYPES:")
print(titanic.dtypes)

# ------------------------------------------
# Display Shape
# ------------------------------------------

print("\nDATASET SHAPE:")
print(titanic.shape)

# ==========================================
# TASK 2
# Descriptive Statistics
# ==========================================

print("\nDESCRIPTIVE STATISTICS")

numeric_cols = titanic.select_dtypes(include=['number'])

print("\nMEAN")
print(numeric_cols.mean())

print("\nMEDIAN")
print(numeric_cols.median())

print("\nSTANDARD DEVIATION")
print(numeric_cols.std())

print("\nMINIMUM")
print(numeric_cols.min())

print("\nMAXIMUM")
print(numeric_cols.max())

print("\nSKEWNESS")
print(numeric_cols.skew())

print("\nKURTOSIS")
print(numeric_cols.kurt())

# ==========================================
# TASK 3
# Histogram of Passenger Ages
# ==========================================

plt.figure(figsize=(8,5))

titanic['Age'].hist(
    bins=20,
    edgecolor='black'
)

plt.title("Titanic Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()

# ==========================================
# TASK 4
# Survival Count by Passenger Class
# ==========================================

plt.figure(figsize=(8,5))

survival_by_class = titanic.groupby('Pclass')['Survived'].sum()

survival_by_class.plot(
    kind='bar'
)

plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Survivors")

plt.show()

# ==========================================
# PART B — SPOTIFY DATASET
# =========================================
# ==========================================
# END OF EXERCISE 1
# ==========================================
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

shipping = pd.read_csv("/content/Train.csv")

print("First 5 Rows")
print(shipping.head())

print("\nMissing Values")
print(shipping.isnull().sum())

corr_matrix = shipping.corr(numeric_only=True)

print("\nCorrelation with Delivery Status")
print(corr_matrix["Reached.on.Time_Y.N"])

plt.figure(figsize=(10,6))

sns.countplot(
    x="Product_importance",
    hue="Warehouse_block",
    data=shipping
)

plt.title(
    "Delivery Status by Product Importance and Warehouse"
)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

# =========================
# HEART DISEASE DATASET
# =========================

heart = pd.read_csv("/content/heart_disease_uci.csv")

print(heart.isnull().sum())

heart = heart.fillna(heart.mean(numeric_only=True))

corr_matrix = heart.corr(numeric_only=True)

plt.figure(figsize=(10,8))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Matrix")
plt.show()

corr_target = corr_matrix["num"]

top5 = corr_target.abs().sort_values(
    ascending=False
)

print("Top 5 Features:")
print(top5.head(6))

fig, axes = plt.subplots(
    1,
    2,
    figsize=(12,5)
)

sns.boxplot(
    x="num",
    y="chol",
    data=heart,
    ax=axes[0]
)

sns.boxplot(
    x="num",
    y="age",
    data=heart,
    ax=axes[1]
)

axes[0].set_title(
    "Cholesterol vs Heart Disease"
)

axes[1].set_title(
    "Age vs Heart Disease"
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# PART C — WORLD HAPPINESS DATASET
# ==========================================
print("\n" + "=" * 50)
print("WORLD HAPPINESS DATASET (2015)")
print("=" * 50)

# Load World Happiness Report (2015) dataset
happiness = pd.read_csv("/content/2015.csv")

print("\nFIRST 5 ROWS:")
print(happiness.head())

# ==========================================
# TASK 6
# Top 20 Happiest Countries
# ==========================================

# Change column names if needed
# Example:
# Country column = Country
# Happiness score column = Score

top20 = happiness.sort_values(
    by='Happiness Score',
    ascending=False
).head(20)

plt.figure(figsize=(12,6))

sns.barplot(
    data=top20,
    x='Country',
    y='Happiness Score'
)

plt.title("Top 20 Happiest Countries")
plt.xlabel("Country")
plt.ylabel("Happiness Score")

plt.xticks(rotation=90)

plt.show()

# ==========================================
# END OF EXERCISE 1
# ==========================================
```
