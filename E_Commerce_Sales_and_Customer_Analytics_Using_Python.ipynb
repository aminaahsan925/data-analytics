```python
import pandas as pd

df = pd.read_csv("/content/Sample - Superstore.csv (1).zip", encoding='latin1')
df.head()
df.info()
df.shape
df.describe()
df.isnull().sum()
df.duplicated().sum()
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
total_sales = df['Sales'].sum()
print(total_sales)
total_profit =df['Profit'].sum()
print(total_profit)
total_salesbyyear=df.groupby('Year')['Sales'].sum()
print(total_salesbyyear)
total_profitbyyear=df.groupby('Year')['Profit'].sum()
print(total_profitbyyear)
import matplotlib.pyplot as plt

df.groupby('Year')['Sales'].sum().plot(kind='bar')
plt.show()
df.groupby('Region')['Sales'].sum()
df.groupby('Region')['Profit'].sum().plot(kind='bar')
plt.show()
df.groupby('Region')['Sales'].sum().plot(kind='bar')
plt.show()
df.groupby('Category')['Sales'].sum()
df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
most_demanding_product=df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(most_demanding_product)
df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)
customerby_profit =df.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False).head(10)
print(customerby_profit)
core =df[['Discount','Profit']].corr()
print(core)
plt.scatter(df['Discount'], df['Profit'])
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.show()
customer_sales = df.groupby('Customer Name')['Sales'].sum()
print(customer_sales)
customer_segments = pd.qcut(
    customer_sales,
    q=3,
    labels=['Low','Medium','High']
)
df.to_csv("superstore_cleaned.csv", index=False)
```
